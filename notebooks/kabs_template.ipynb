```python
import taichi as ti
import porespy as ps
import matplotlib.pyplot as plt
from kabs import compute_permeability, solve_flow
```

```python
# Tell taichi how to run
# ti.init(arch=ti.metal)\
ti.init(arch=ti.cpu)
```

```python
# Generate a test image
im = ps.generators.cylinders([200, 200, 100], r=10, porosity=0.7, seed=1)
```

```python
ax = "x"
soln = solve_flow(im=im, direction=ax, tol=1e-3)
compute_permeability(soln, direction=ax)
```

```python
ax = "y"
soln = solve_flow(im=im, direction=ax, tol=1e-3)
compute_permeability(soln, direction=ax)
```

```python
ax = "z"
soln = solve_flow(im=im, direction=ax, tol=1e-3)
compute_permeability(soln, direction=ax)
```
